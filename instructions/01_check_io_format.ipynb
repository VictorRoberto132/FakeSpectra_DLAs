## In this notebook I explain how can we read and understand .pkl format since we use this a lot here to store the numbers from runs

```python
#Pikles is a package from pandas, so we need to import pandas
import pandas as pd
import os
import sys
```

```python
# We first have to set the pathof the file we want to read: 
test_dir = "/home/vroberto/FakeSpectra_DLAs/data_TNG50-4_snap025_5runs"
```

```python
# Now that we have a directory we can set the path of the file we want to read.
file_path = os.path.join(test_dir, "TNG50-4_z3_637Halos_HCDMode_Stratified_run0_seed42.pkl")
df = pd.read_pickle(file_path)
```

```python
# If we want to read the information of the dataframe we can use the info() method:
df.info()
```

```python
# If we want to see the first columns, we can use the head() method:
df.head()
```

```python
# If we want to se the first row of the dataframe we can use the iloc method:
df.iloc[0]
```

```python
# The first temperature array: 
print(df["temp"].iloc[0])
```
